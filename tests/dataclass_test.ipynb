```python
from pmrf.models import Inductor
import skrf as rf
import matplotlib.pyplot as plt

from pmrf._math import dB20
from pmrf import Frequency
```

```python
L = 1.0e-9
skrf_freq = rf.Frequency(50, 200, 151, 'MHz')
prf_freq = Frequency(skrf_freq)
```

```python

skrf_inductor = rf.media.DefinedGammaZ0(skrf_freq).inductor(L)
skrf_s11 = skrf_inductor.s[:, 0, 0]
```

```python
prf_inductor = Inductor(L)
prf_inductor.to_skrf(skrf_freq)
```

```python
plt.plot(skrf_freq.f_scaled, dB20(skrf_s11))
plt.plot(skrf_freq.f_scaled, dB20(prf_s11))
```
